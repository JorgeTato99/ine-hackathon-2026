```python
from pathlib import Path
import sys
import polars as pl
from ine_hackathon_2026.ccpc import validate_municipal

project_root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'data/reference/municipios_para_mapa.csv').is_file())
print(sys.version)
municipal = pl.read_csv(project_root / 'data/reference/municipios_para_mapa.csv', schema_overrides={k: pl.String for k in ['cvegeo','clave_ine','cve_ent_ine','cve_mun_ine']})
validate_municipal(municipal)
municipal.shape
```

```python
municipal.group_by('anio').agg(pl.len().alias('municipios')).sort('anio')
```
